```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolo26n.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data="signature.yaml", epochs=100, imgsz=640)
```

```python
from ultralytics import YOLO

# Load a model
model = YOLO("/content/runs/detect/train2/weights/best.pt")  # load a signature-detection fine-tuned model

# Inference using the model
results = model.predict("https://ultralytics.com/assets/signature-s.mp4", conf=0.75, save=True)
print(results[0])
for result in results:
    if result.probs is not None:
        print("Classification probabilities:", result.probs)
        print(result)
```
